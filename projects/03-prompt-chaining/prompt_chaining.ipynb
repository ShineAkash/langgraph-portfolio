```python
from langgraph.graph import StateGraph,START,END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
model =ChatGroq(model=os.environ["GROQ_MODEL"])
```

```python
class BlogState(TypedDict):
    title :str
    outline :str
    content: str
```

```python
def generate_outline(state:BlogState)->BlogState:
    title=state['title']
    prompt=f"Generate a detailed outline for a blog post on the topic: {title}"
    outline=model.invoke(prompt).content
    state['outline']=outline
    return state
```

```python
def generate_blog(state:BlogState)->BlogState:
    title=state['title']
    outline=state['outline']
    prompt=f"Write a detailed blog on the title:{title} with the following outline:\n {outline}"
    content=model.invoke(prompt).content
    state['content']=content
    return state 
```

```python
graph=StateGraph(BlogState)
graph.add_node('generate_outline',generate_outline)
graph.add_node('generate_blog',generate_blog)

graph.add_edge(START,'generate_outline')
graph.add_edge('generate_outline','generate_blog')
graph.add_edge('generate_blog',END)
workflow=graph.compile()
```

```python
initail_state={"title":"The future of AI in healthcare"}
final_state=workflow.invoke(initail_state)
print(final_state['content'])
```
